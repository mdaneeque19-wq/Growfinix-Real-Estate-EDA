```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries are ready!")
```

```python
data = {
    "Neighborhood": [
        "Anna Nagar", "Velachery", "Tambaram",
        "Anna Nagar", "Velachery", "Tambaram",
        "Anna Nagar", "Velachery", "Tambaram",
        "Anna Nagar"
    ],
    "Size_sqft": [
        1200, 950, 800, 1500, 1100,
        900, 1800, 1000, 750, 1300
    ],
    "Bedrooms": [3, 2, 2, 3, 2, 2, 4, 2, 1, 3],
    "Price_lakhs": [
        85, 60, 40, 110, 70,
        45, 150, 65, 35, 90
    ]
}

df = pd.DataFrame(data)

print("Real Estate Dataset")
display(df)
```

```python
print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("Statistical Summary:")
display(df.describe())

plt.figure(figsize=(8, 5))
sns.boxplot(x=df["Price_lakhs"])

plt.title("Boxplot of Property Prices")
plt.xlabel("Price (Lakhs)")
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=df,
    x="Neighborhood",
    y="Price_lakhs",
    errorbar=None
)

plt.title("Average Property Price by Neighborhood")
plt.xlabel("Neighborhood")
plt.ylabel("Price (Lakhs)")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Size_sqft",
    y="Price_lakhs",
    hue="Neighborhood",
    s=100
)

plt.title("Property Size vs. Price")
plt.xlabel("Property Size (sq. ft.)")
plt.ylabel("Price (Lakhs)")
plt.show()
```

```python
# Create a practice copy of our dataset
clean_df = df.copy()

# Add a missing value for practice
clean_df.loc[0, "Price_lakhs"] = np.nan

print("Missing values before cleaning:")
print(clean_df.isnull().sum())

# Fill missing prices with the median price
clean_df["Price_lakhs"] = clean_df["Price_lakhs"].fillna(
    clean_df["Price_lakhs"].median()
)

print("\nMissing values after cleaning:")
print(clean_df.isnull().sum())

print("\nCleaned dataset:")
display(clean_df)
```

```python
Q1 = df["Price_lakhs"].quantile(0.25)
Q3 = df["Price_lakhs"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df["Price_lakhs"] < lower_limit) |
    (df["Price_lakhs"] > upper_limit)
]

print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
print("\nPotential outliers:")
display(outliers)
```

```python
print("GROWFINIX - REAL ESTATE EDA")
print("-" * 35)

print("Total properties:", len(df))
print("Total neighborhoods:", df["Neighborhood"].nunique())
print("Average property price:", df["Price_lakhs"].mean(), "lakhs")
print("Minimum property price:", df["Price_lakhs"].min(), "lakhs")
print("Maximum property price:", df["Price_lakhs"].max(), "lakhs")
print("Average property size:", df["Size_sqft"].mean(), "sq. ft.")
print("Potential outliers detected:", len(outliers))

print("\nEDA completed successfully!")
```
